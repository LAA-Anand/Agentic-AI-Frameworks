```python
import os
from getpass import getpass

# Enter OpenAI API key securely
api_key = getpass("Enter your OpenAI API key: ")

# Set environment variables
os.environ["OPENAI_API_KEY"] = api_key
os.environ["OPENAI_MODEL_NAME"] = "gpt-4o"
```

```python
# Warning control
import warnings
warnings.filterwarnings('ignore')
```

```python
from agno.agent import Agent
from agno.models.openai import OpenAIResponses
from agno.tools.openai import OpenAITools
from agno.tools.duckduckgo import DuckDuckGoTools
from agno.tools.websearch import WebSearchTools
from agno.tools.dalle import DalleTools
from agno.team import Team
from textwrap import dedent
```

web_agent – Searches the Web, writer_agent – Writes the Article, image_agent – Creates Visuals

```python
web_agent = Agent(
    name="Web Agent",
    role="Search the web for information on Taj Mahal India",
    model=OpenAIChat(id="gpt-4o"),
    tools=[DuckDuckGoTools()],
    instructions="Give historical information",
    markdown=True,
)

writer_agent = Agent(
    name="Writer Agent",
    role="Write comprehensive article on the provided topic",
    model=OpenAIChat(id="gpt-4o"),
    tools=[WebSearchTools()],
    instructions="Use outlines to write articles",
    markdown=True,
)

image_agent = Agent(
    name="Image Generation Agent",
    model=OpenAIResponses(id="gpt-5.2"),
    tools=[
        OpenAITools(image_model="gpt-image-2")
    ],
    instructions=[
        "You are an expert AI artist.",
        "Generate images when requested.",
        "Create visually appealing and historically accurate illustrations."
    ],
    markdown=True
)
```

```python
agent_team = Team(
    name="Historical Research Team",
    members=[
        web_agent,
        writer_agent,
        image_agent
    ],
    model=OpenAIChat(id="gpt-4o-mini"),
    instructions=[
        "Research the requested historical topic.",
        "Write a comprehensive article.",
        "Generate a relevant image.",
        "Present the final article and image."
    ],
    markdown=True
)
```

```python
response = agent_team.run(
    "Write a comprehensive article on India's Taj Mahal "
    "and generate a historically appropriate image."
)

print(response.content)
```

```python
image_response = image_agent.run(
    "Generate a historically accurate image of the Taj Mahal "
    "with its gardens, reflecting pool and four minarets."
)

print(image_response)
```
