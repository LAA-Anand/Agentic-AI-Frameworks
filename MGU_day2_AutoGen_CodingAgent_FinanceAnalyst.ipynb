```python
import os
```

```python
api_key = getpass("Enter your OpenAI API key: ")

os.environ["OPENAI_API_KEY"] = api_key
```

```python
llm_config={"model": "gpt-4o"}
```

```python
from autogen.coding import LocalCommandLineCodeExecutor
```

The LocalCommandLineCodeExecutor class from the autogen.coding module in the AutoGen framework is designed to execute code snippets locally on your machine or environment’s command line (shell). It provides a controlled environment for running code generated or requested by an AI agent as part of a coding or reasoning workflow.

```python
executor = LocalCommandLineCodeExecutor(
    timeout=60, #If the code takes longer, it will be terminated to prevent hanging or infinite loops.
    work_dir="coding",
)
```

Create agents

```python
from autogen import ConversableAgent, AssistantAgent
```

Agent with code executor configuration

```python
code_executor_agent = ConversableAgent(
    name="code_executor_agent",
    llm_config=False, #This means the agent is not directly bound to an LLM model in this config.
    code_execution_config={"executor": executor}, #This is the key to enabling code execution inside the agent.
    human_input_mode="ALWAYS", #This means the agent always waits for human input before proceeding.
    default_auto_reply=
    "Please continue. If everything is done, reply 'TERMINATE'.",
)
```

Agent with code writing capability

```python
#The writer agent is focused solely on producing code or text via the LLM
code_writer_agent = AssistantAgent(
    name="code_writer_agent",
    llm_config=llm_config,
    code_execution_config=False, #the agent does not execute any code
    human_input_mode="NEVER", #the agent runs autonomously without expecting human input after initialization.
)
```

```python
code_writer_agent_system_message = code_writer_agent.system_message
```

```python
print(code_writer_agent_system_message)
```

Ask the two agents to collaborate on a stock analysis task

```python
import datetime
#Stock gain YTD" means the percentage change in a stock's price from the beginning of the current year (January 1st) up to the current date.
today = datetime.datetime.now().date()
message = f"Today is {today}. "\
"Create a plot showing stock gain YTD for Tata ELXI and Infosys "\
"Make sure the code is in markdown code block and save the figure"\
" to a file ytd_stock_gains.png."""
```

```python
chat_result = code_executor_agent.initiate_chat(
    code_writer_agent,
    message=message,
)
```
